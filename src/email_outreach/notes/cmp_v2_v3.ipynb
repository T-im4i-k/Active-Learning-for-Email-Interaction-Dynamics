```python
import pandas as pd

from email_outreach.experiment_utils.metric_loader import load_experiment_metrics
from email_outreach.ml.shallow_autoencoder.abstract_contextual_model import AbstractContextualModel
from email_outreach.utils.constants import ROOT_FOLDER
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
```

```python
def get_precisions_recalls(path: Path, quantiles=1 - np.linspace(start=0, stop=1, num=100)):
    experiments: list[list] = load_experiment_metrics(path)
    recalls: list[float] = []
    precisions: list[float] = []
    for experiment in experiments:
        recall, precision = AbstractContextualModel.calculate_recall_precision_at_quantiles(experiment,
                                                                                            quantiles=quantiles)[1:3]

        recalls.append(recall)
        precisions.append(precision)

    return recalls, precisions
```

### Geometric

```python
geometric_alpha_path_v1: Path = ROOT_FOLDER / "experiments" / "test_set" / "1" / "contextual_bandit" / "geometric_alpha_v1"

geometric_alpha_path_v2: Path = ROOT_FOLDER / "experiments" / "test_set" / "1" / "contextual_bandit" / "geometric_alpha_v2"
```

```python
geometric_alpha_recalls_v1, geometric_alpha_precisions_v1 = get_precisions_recalls(geometric_alpha_path_v1)
geometric_alpha_recalls_v2, geometric_alpha_precisions_v2 = get_precisions_recalls(geometric_alpha_path_v2)
```

```python
plt.figure(figsize=(15, 15))

plt.plot(np.linspace(start=0, stop=1, num=100)[:26], np.array(geometric_alpha_recalls_v1).mean(axis=0)[:26],
         label="Geometric Alpha Recalls v1")
plt.plot(np.linspace(start=0, stop=1, num=100)[:26], np.array(geometric_alpha_recalls_v2).mean(axis=0)[:26],
         label="Geometric Alpha Recalls v2")

plt.legend()
```
