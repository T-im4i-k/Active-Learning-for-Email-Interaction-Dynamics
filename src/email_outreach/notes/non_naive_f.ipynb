```python
import pandas as pd

from email_outreach.experiment_utils.metric_loader import load_experiment_metrics
from email_outreach.ml.shallow_autoencoder.abstract_contextual_model import AbstractContextualModel
from email_outreach.utils.constants import ROOT_FOLDER
from pathlib import Path
import numpy as np
import matplotlib.pyplot as plt
```

```python
def get_precisions_recalls(path: Path, quantiles=1 - np.linspace(start=0, stop=1, num=100)):
    experiments: list[list] = load_experiment_metrics(path)
    recalls: list[float] = []
    precisions: list[float] = []
    for experiment in experiments:
        recall, precision = AbstractContextualModel.calculate_recall_precision_at_quantiles(experiment,
                                                                                            quantiles=quantiles)[1:3]

        recalls.append(recall)
        precisions.append(precision)

    return recalls, precisions
```

```python
naive_f_path: Path = ROOT_FOLDER / "experiments" / "test_set" / "1" / "contextual_bandit" / "baseline_v2"
non_naive_path: Path = ROOT_FOLDER / "experiments" / "test_set" / "1" / "contextual_bandit" / "non_naive_f_v2"
non_naive_f_path_v1: Path = ROOT_FOLDER / "experiments" / "test_set" / "1" / "contextual_bandit" / "non_naive_f_v1"
```

```python
naive_recalls, naive_precisions = get_precisions_recalls(naive_f_path)
non_naive_recalls, non_naive_precisions = get_precisions_recalls(non_naive_path)
non_naive_recalls_v1, non_naive_precisions_v1 = get_precisions_recalls(non_naive_f_path_v1)
```

```python
print(f"Naive recall: {np.array(naive_recalls).mean(axis=0)}")
print(f"Non-naive recall: {np.array(non_naive_recalls).mean(axis=0)}")
```

```python
plt.figure(figsize=(15, 15))

plt.plot(np.linspace(start=0, stop=1, num=100), np.array(naive_recalls).mean(axis=0), label="Naive f Recalls")
plt.plot(np.linspace(start=0, stop=1, num=100), np.array(non_naive_recalls).mean(axis=0),
         label="Non Naive f Alpha Recalls")

plt.plot(np.linspace(start=0, stop=1, num=100), np.array(non_naive_recalls_v1).mean(axis=0),
         label="Non Naive f Alpha Recalls v1")


plt.vlines(x=0.15, ymin=0, ymax=1)
plt.legend()
plt.show()
```
