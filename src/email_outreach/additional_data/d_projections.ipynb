```python
from sklearn.manifold import TSNE
import numpy as np
import matplotlib.pyplot as plt
import umap
```

```python
D_pos = np.load("proj/sae_D.npy")
D_neg = np.load("proj/neg_sae_D.npy")
```

```python
D_pos.shape, D_neg.shape
```

```python
perplexity = [20, 40]
tsne_pos_list = []
tsne_neg_list = []

for p in perplexity:
    pos_tsne = TSNE(perplexity= p ,n_components=2, random_state=42, n_jobs=2)
    neg_tsne = TSNE(perplexity= p ,n_components=2, random_state=42, n_jobs=2)

    pos_tsne.fit(D_pos)
    neg_tsne.fit(D_neg)

    tsne_pos_list.append(pos_tsne)
    tsne_neg_list.append(neg_tsne)
```

```python
em_pos_20 = tsne_pos_list[0].embedding_

plt.scatter(em_pos_20[:, 0], em_pos_20[:, 1], s=0.1)
```

```python
em_pos_40 = tsne_pos_list[2].embedding_

plt.scatter(em_pos_40[:, 0], em_pos_40[:, 1], s=0.1)
```

```python
em_neg_20 = tsne_pos_list[1].embedding_

plt.scatter(em_neg_20[:, 0], em_neg_20[:, 1], s=0.1)
```

```python
em_neg_40 = tsne_pos_list[3].embedding_

plt.scatter(em_neg_40[:, 0], em_neg_40[:, 1], s=0.1)
```
