```python
from sklearn.manifold import TSNE
import numpy as np
import matplotlib.pyplot as plt
import umap
```

```python
D_pos = np.load("D/sae_D.npy")
D_neg = np.load("D/neg_sae_D.npy")
```

```python
D_pos.shape, D_neg.shape
```

```python
tsne_pos = TSNE(n_components=2, random_state=42, n_jobs=2)
tsne_neg = TSNE(n_components=2, random_state=42, n_jobs=2)
```

```python
D_pos_tsne = tsne_pos.fit_transform(D_pos)
D_neg_tsne = tsne_neg.fit_transform(D_neg)
```

```python
plt.scatter(D_neg_tsne[:, 0], D_neg_tsne[:, 1], s=0.1)
```

```python
plt.scatter(D_pos_tsne[:, 0], D_pos_tsne[:, 1], s=0.1)
```

```python
umap_pos = umap.UMAP(n_components=2, random_state=42)
umap_neg = umap.UMAP(n_components=2, random_state=42)
```

```python
D_pos_umap = umap_pos.fit_transform(D_pos)
```

```python
D_neg_umap = umap_neg.fit_transform(D_neg)
```

```python
plt.scatter(D_neg_umap[:, 0], D_neg_umap[:, 1], s=0.1)
```

```python
plt.scatter(D_pos_umap[:, 0], D_pos_umap[:, 1], s=0.1)
```
